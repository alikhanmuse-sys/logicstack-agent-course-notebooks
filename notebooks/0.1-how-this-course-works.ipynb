## The idea in plain words

Most courses about AI agents teach you to watch someone else build things. This one is built the other way round. Every lesson ends with code you wrote and ran yourself. Every module ends with a working project you can show to someone. The certificate at the end is not a prize for finishing lessons. It is proof you can build and debug an agent under exam conditions.

There are 7 modules, numbered 0 to 6, plus a capstone project and a final exam:

| Module | What you build |
|---|---|
| 0 — Orientation & Setup | A working dev environment and your first LLM call |
| 1 — LangChain Core | A CLI assistant with memory and structured output |
| 2 — Tools & Agents | A research assistant agent, then your own task agent |
| 3 — RAG | A docs Q&A bot with citations |
| 4 — LangGraph Foundations | A customer-support triage graph |
| 5 — Advanced LangGraph | A multi-agent research team |
| 6 — Production | Your capstone deployed behind an API |

**Capstone (choose one track):** an autonomous research agent with cited sources, a support agent with human approval steps, or a document intelligence pipeline. Your capstone must use custom tools, RAG, a multi-step LangGraph workflow, an eval set with results, a deployed demo, and a README.

**Final exam (gates the certificate):** Part A is practical and time-boxed: build or debug a small agent task against a hidden test. Part B is theory: questions on agent patterns, failure modes, evals, safety, and cost. Pass mark is 70%. You only get the certificate if you pass. Details like retake rules are decided before the course goes live.

Each lesson follows the same shape: learning objectives, a plain explanation, runnable code with pinned versions, a "What can go wrong" section, and a short quiz with answers. The "what can go wrong" sections are not filler. They are the most useful part of the course. Agents fail in specific, repeatable ways, and learning those ways now saves you real debugging time later.

## Why LangChain first, then LangGraph

LangChain and LangGraph are not rivals. They are layers of the same stack, built by the same team. LangChain gives you a ready-made agent loop: give it a model and some tools and it runs. LangGraph gives you the engine underneath: nodes, edges, and state that you control yourself.

The course starts with LangChain because a ready-made loop teaches you what an agent *does* before you learn how to *build the loop yourself*. In Module 4 you will hit real limits of the simple loop: things that need branching, retries with custom logic, or a human approval step. That is the moment graphs stop being abstract and start being the obvious tool. You will rebuild your Module 2 agent as a graph and compare the two side by side.

One honest warning, echoed by working developers: do not start at LangGraph because it feels more serious. People who do usually spend weeks rebuilding what the ready-made agent loop gives them for free. Start high level, drop down only when you have a concrete reason.

## A note on Europe

This course is written for a European audience. Where it matters, lessons call out the EU angle explicitly: running models locally with Ollama keeps data inside your own machine (data residency), agent logs can contain personal data and fall under GDPR, and deploying agents makes you a "deployer" under the EU AI Act with real obligations. These are not side notes. They change real engineering decisions, and Module 6 covers them properly.

## What can go wrong

- **Skipping the setup lesson.** Almost every "LangChain is broken" story on forums starts with a broken Python environment, not a broken library. Run the readiness check above and Module 0.2 carefully.
- **Version drift.** LangChain and LangGraph release often. Every lesson pins the versions it was written against. If code from a lesson stops working six months from now, check the pinned versions first.
- **Tutorial-only learning.** Reading every lesson without running the code teaches you vocabulary, not skill. The exam tests skill. Type the code out yourself; do not copy-paste it.
- **Chasing the newest model.** Lesson code uses stable, cheap models. A newer model does not make a broken agent design work. Fix the design first.
